# Number Guessing Game

## Asses the guessed number with the secret number

```python
secret_number = 7
for i in range(0,7):
    guessed_value = int(input("Enter your guess"))
    if guessed_value == secret_number:
        print(f"Your guessed value {guessed_value} is right")
        break
    else:
        print(f"Your guessed value {guessed_value} is wrong")
```

# Using while loop

```python
secret_number = 7
while True:
    guessed_value = int(input("Enter your guess"))
    if guessed_value == secret_number:
        print(f"Your guessed value {guessed_value} is right")
        break
    else:
        print(f"Your guessed value {guessed_value} is wrong")
```
