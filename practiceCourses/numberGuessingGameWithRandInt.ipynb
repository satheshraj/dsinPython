# Number Guessing Game

### Asses the guessed number with the randint

```python
import random

secret_number = random.randint(0,100)
for i in range(1,8):
    print(f"Attempt :{i}")
    guessed_value = int(input("Enter your guess"))
    if guessed_value < 0 or guessed_value > 100:
        print("Please enter guesses between 0 and 100")
    elif guessed_value == secret_number:
        print(f"Your guessed value {guessed_value} is right. Found in attempt no {i}")
        break
    else:
        print(f"Your guessed value {guessed_value} is wrong")
        if i == 7:
            print(f"You have reached the max number of attempts. The secret number was {secret_number}")
```
